```python
import numpy as np
import matplotlib.pyplot as plt

# 生成数据
x = np.linspace(0, 4 * np.pi, 200)
y = np.sin(x) * np.exp(-0.1 * x)

# 创建图形
plt.figure(figsize=(8, 4))
plt.plot(x, y, color='tab:purple', linewidth=2)
plt.fill_between(x, y, 0, color='tab:purple', alpha=0.2)

plt.title('阻尼正弦波', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)

plt.show()
```
