```python
"""查看本机 GPU / 显卡型号。

依次尝试三种方式：
1. Windows WMI/PowerShell —— 所有显卡（独显 + 集显，含 AMD/Intel）
2. nvidia-smi —— NVIDIA 显卡详细信息（显存、驱动版本）
3. PyTorch —— CUDA 可用性与设备名
"""

import platform
import shutil
import subprocess

print("系统:", platform.system(), platform.release())


# ---------------------------------------------------------- 1) Windows 全部显卡
if platform.system() == "Windows":
    try:
        out = subprocess.run(
            ["powershell", "-NoProfile", "-Command",
             "Get-CimInstance Win32_VideoController | "
             "Select-Object Name, DriverVersion | Format-List"],
            capture_output=True, text=True, timeout=20,
            encoding="gbk", errors="ignore",
        )
        print("\n[Windows] 检测到的显卡：")
        print(out.stdout.strip())
    except Exception as e:
        print("\nWMI 检测失败:", e)


# ---------------------------------------------------------- 2) NVIDIA 显卡
if shutil.which("nvidia-smi"):
    try:
        out = subprocess.run(
            ["nvidia-smi",
             "--query-gpu=name,memory.total,driver_version",
             "--format=csv,noheader"],
            capture_output=True, text=True, timeout=15,
        )
        print("\n[NVIDIA] nvidia-smi 检测到：")
        for line in out.stdout.strip().splitlines():
            print("  ", line)
    except Exception as e:
        print("\nnvidia-smi 调用失败:", e)
else:
    print("\n未找到 nvidia-smi（可能不是 NVIDIA 显卡，或未安装驱动）")


# ---------------------------------------------------------- 3) PyTorch CUDA
try:
    import torch
    print("\n[PyTorch] CUDA 可用:", torch.cuda.is_available())
    if torch.cuda.is_available():
        for i in range(torch.cuda.device_count()):
            print(f"  GPU {i}: {torch.cuda.get_device_name(i)}")
except ImportError:
    print("\n未安装 PyTorch，跳过 CUDA 检测")
```
